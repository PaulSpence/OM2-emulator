# Century-scale rollouts: flux-form, stable, stochastic emulator (placeholder)

Placeholder for a new emulator aimed at **reproducible rollouts over centuries**. Nothing here is implemented yet.

**Goal:** over hundreds of years, reproduce the *statistics* of ACCESS-OM2 (mean state, variance, spectra, global budgets), not a single trajectory.

**Why a new design:** the UNet emulator in `Ocean_Emulator.ipynb` is accurate at short leads but develops grid-scale striping that grows by roughly 8% per monthly step, because nothing in its structure prevents growth or spurious sources and sinks.

## Proposed design

1. **Flux form.** The network predicts horizontal heat and freshwater transports across cell faces. A fixed finite-volume step applies `ΔC = dt · (Q_surface − ∇·F)`, so the global budget follows the surface flux exactly by construction.
2. **Stable by construction.** Transport = advection by a learned divergence-free flow (from a streamfunction, which conserves anomaly variance) + diffusion with a learned positive diffusivity (which only damps). Without forcing, anomaly variance cannot grow. The network runs once a month; a cheap stencil solver sub-steps within the month to keep the transport numerically stable.
3. **Stochastic residual ("neural Hasselmann model").** A learned state-dependent noise term supplies the variability the deterministic core averages away. Trained with a probabilistic score (CRPS / energy score) over multi-year windows.
4. **Slow core, fast fields computed.** Only OHC and freshwater content are fed back step to step. Fast or non-conserved fields (eta_t, surface speed) are computed each month from the slow state + forcing and never fed back.
